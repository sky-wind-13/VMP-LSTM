```python
import pandas as pd
import numpy as np
from scipy import interpolate

# 读取csv文件
df = pd.read_csv('D:\\study\\每天更新\\船舶轨迹可视化\\数据预处理\\插值可视化 -305190000前.csv', encoding='utf-8')

# 根据MMSI进行分组
grouped = df.groupby('MMSI')

# 新的数据保存到一个新的csv文件
new_df = pd.DataFrame(columns=['BaseDateTime', 'MMSI', 'LAT', 'LON'])

for name, group in grouped:
    if len(group) < 4:  # 如果在这个时间段内记录的ais数据记录小于4条，直接跳过
        continue
    
    # 获取每个组的LAT,LON,TIME列
    lat = group['LAT'].values
    lon = group['LON'].values
    time = group['BaseDateTime'].values
    
    # 根据TIME列生成整数分钟的time,除1000因为时间戳数据格式错误，多了三个零
    time_int = np.unique(np.floor(time))
    
    # 移除重复的时间戳，并同步移除lat和lon中的重复值
    unique_indices = np.unique(np.floor(time), return_index=True)[1]
    lat = lat[unique_indices]
    lon = lon[unique_indices]
    
    # 插值
    # f_lat = interpolate.interp1d(time_int, lat, kind='cubic')
    # f_lon = interpolate.interp1d(time_int, lon, kind='cubic')

    f_lat = interpolate.interp1d(time_int, lat, kind='linear')
    f_lon = interpolate.interp1d(time_int, lon, kind='linear')
    # f_lat = interpolate.interp1d(time_int, lat)
    # f_lon = interpolate.interp1d(time_int, lon)


    # 生成每隔10秒的时刻,保证时间戳是10的倍数
    new_time = np.arange(time_int.min() + (10 - time_int.min() % 10), time_int.max() - (time_int.max() % 10), 10)

    # 计算插值结果
    new_lat = f_lat(new_time)
    new_lon = f_lon(new_time)

    # 保存到新的dataframe中
    temp_df = pd.DataFrame({'BaseDateTime': new_time, 'MMSI': name, 'LAT': new_lat, 'LON': new_lon})
    new_df = pd.concat([new_df, temp_df])

# 保存到csv文件
new_df.to_csv('插值可视化 -305190000后线性插值.csv', index=False)
# print(new_df.info())
new_df.info()
```

```python
import pandas as pd
import numpy as np
from scipy import interpolate

# 读取csv文件
df = pd.read_csv('D:\study\每天更新\船舶轨迹可视化\数据预处理_高伟\\04_aisdk_2023_04_13_val.csv',encoding='utf-8')

# 根据MMSI进行分组
grouped = df.groupby('MMSI')

# 新的数据保存到一个新的csv文件
new_df = pd.DataFrame(columns=['Timestamp','MMSI', 'Latitude', 'Longitude'])

for name, group in grouped:
    if len(group)<4:# 如果在这个时间段内记录的ais数据记录小于4条，直接跳过
        continue
    
    # 获取每个组的LAT,LON,TIME列
    lat = group['Latitude'].values
    lon = group['Longitude'].values
    time = group['Timestamp'].values
    
    # 根据TIME列生成整数分钟的time,除1000因为时间戳数据格式错误，多了三个零
    time_int = np.unique(np.floor(time))

    # 插值
    f_lat = interpolate.interp1d(time_int, lat,kind='cubic')
    f_lon = interpolate.interp1d(time_int, lon,kind='cubic')

    # 生成每隔10秒的时刻,保证时间戳是10的倍数
    new_time = np.arange(time_int.min()+(10-time_int.min()%10), time_int.max()-(time_int.max()%10), 10)

    # 计算插值结果
    new_lat = f_lat(new_time)
    new_lon = f_lon(new_time)

    # 保存到新的dataframe中
    temp_df = pd.DataFrame({'Timestamp': new_time, 'MMSI': name, 'Latitude': new_lat, 'Longitude': new_lon})
    new_df = pd.concat([new_df, temp_df])

# 保存到csv文件
new_df.to_csv('05_aisdk_2023_04_13_train_cubic.csv', index=False)
new_df.info
```
