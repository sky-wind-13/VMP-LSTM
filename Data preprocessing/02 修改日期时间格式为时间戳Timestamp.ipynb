### train

```python
import pandas as pd

# 读取CSV文件
data = pd.read_csv("D:\\study\\每天更新\\船舶轨迹可视化\\数据预处理\\路易斯安那1_时筛_位筛_逆_起.csv")

# 将Time列转换为时间戳格式
data["Timestamp"] = (pd.to_datetime(data["Timestamp"]).astype('int64') // 10**9).astype('int32')

# 选择需要保存的列（MMSI、Latitude、Longitude和Timestamp）
new_data = data[[ "Timestamp","MMSI", "Latitude", "Longitude"]]

# 将数据保存到新的CSV文件中
new_data.to_csv("02_路易斯安那1_时筛_位筛_逆_起.csv", index=False)
new_data.info
```
