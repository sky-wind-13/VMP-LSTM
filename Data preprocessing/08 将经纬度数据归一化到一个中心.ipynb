### train

```python
import pandas as pd
import numpy as np

# 读取csv文件
data = pd.read_csv("07_aisdk_2023_04_13_train.csv")

# 给出 矩形区域的中心点坐标
mean_lon = data["Longitude"].mean()
mean_lat = data["Latitude"].mean()

# 计算矩形区域的宽度和高度（以经纬度差值的绝对值表示）
width = np.abs(data["Longitude"].max() - data["Longitude"].min())
height = np.abs(data["Latitude"].max() - data["Latitude"].min())

# 对经纬度进行归一化（即减去中心点坐标，再除以矩形区域的宽度和高度）
data["Longitude"] = (data["Longitude"] - mean_lon) / width*10
data["Latitude"] = (data["Latitude"] - mean_lat) / height*10

# data = data.drop(labels='Longitude',axis=1)
# data = data.drop(labels='Latitude'.axis=1)

# 保存归一化后的数据到新的csv文件
data.to_csv("08_aisdk_2023_04_13_train.csv", index=False)
```
