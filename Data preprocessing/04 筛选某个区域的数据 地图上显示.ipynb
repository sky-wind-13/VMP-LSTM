```python
import csv

# 构造矩形区域
def rect_area(lat, lon, lat_A, lon_A, width, height):
    if lat <= lat_A and lat >= lat_A - height and lon >= lon_A and lon <= lon_A + width:
        return True
    return False

# 读取csv文件
with open("02_aisdk_2023_04_13_train.csv", "r",encoding='utf-8') as f:
    reader = csv.reader(f)
    headers = next(reader)
    data = [row for row in reader]

# 给定左上经纬度和长宽
lat_A = 55.75261
lon_A = 12.61431

width = 0.1
height = 0.1
filtered_data = [row for row in data if rect_area(float(row[2]), float(row[3]), lat_A, lon_A, width, height)]

# 保存数据到新的csv文件
with open("04_aisdk_2023_04_13_train_del.csv", "w") as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows(filtered_data)
    
    
    
import csv

# 读取csv文件
with open('04_aisdk_2023_04_13_train_del.csv', 'r') as file:
    reader = csv.reader(file)
    lines = [line for line in reader if line]

# 将有数据的行写入一个新的csv文件
with open('04_aisdk_2023_04_13_train.csv', 'w', newline='') as new_file:
    writer = csv.writer(new_file)
    writer.writerows(lines)

    
    
    
# 删除有空行的文件    
import os
os.remove('04_aisdk_2023_04_13_train_del.csv')
    
    
    
import pandas as pd
import folium
import random

# 读取CSV文件
df = pd.read_csv('04_aisdk_2023_04_13_train.csv', usecols=['Timestamp', 'MMSI', 'Latitude', 'Longitude'])

# 根据MMSI进行分组
grouped = df.groupby('MMSI')

# 创建地图
m = folium.Map(location=[df['Latitude'].mean(), df['Longitude'].mean()], zoom_start=10)

# 为每个组添加标记
for name, group in grouped:
    color = '#{:06x}'.format(random.randint(0, 0xFFFFFF))
    for lat, lon, in zip(group['Latitude'], group['Longitude']):
        folium.Marker(location=[lat, lon], icon=folium.Icon(color=color)).add_to(m)

# 保存地图
m.save('04_aisdk_2023_04_13_train.html')
```
