### 00:00-13:59 14个小时 train

```python
import csv
from datetime import datetime

# 读取csv文件
with open('00_aisdk_2023-04-13.csv', 'r') as f:
    reader = csv.reader(f)
    headers = next(reader)
    rows = [row for row in reader]

# 筛选数据,2个小时的数据
start_time = datetime.strptime('2023-04-13 00:00:00', '%Y-%m-%d %H:%M:%S')
end_time = datetime.strptime('2023-04-13 13:59:59', '%Y-%m-%d %H:%M:%S')
filtered_rows = [row for row in rows if start_time <= datetime.strptime(row[0], '%d/%m/%Y %H:%M:%S') <= end_time]


# 写入新的csv文件
with open('01_aisdk_2023_04_13_train_null.csv', 'w') as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows(filtered_rows)
#  筛选的文件存在空行   
    
# ------------------------------------
    
    
#删除空行
import csv

# 读取csv文件
with open('01_aisdk_2023_04_13_train_null.csv', 'r') as file:
    reader = csv.reader(file)
    lines = [line for line in reader if line]

# 将有数据的行写入一个新的csv文件
with open('01_aisdk_2023_04_13_train.csv', 'w', newline='') as new_file:
    writer = csv.writer(new_file)
    writer.writerows(lines)
    
    
    
import os
# 删除原来有空行的文件
os.remove('01_aisdk_2023_04_13_train_null.csv')
```
