## 例子

```python
from math import sin, cos, sqrt, atan2, radians

# 地球半径，单位米
R = 6371000

# A点经纬度坐标
lat_A = 55.75261
lon_A = 12.61431

# B点经纬度坐标
lat_B = 55.65261
lon_B = 12.71431

# 计算A点的经纬度对应的XY坐标
x_a = 0
y_a = 0

# 计算B点的经纬度对应的XY坐标
x_b = R * cos(radians(lat_B - lat_A)) * cos(radians(lon_B - lon_A))
y_b = R * cos(radians(lat_B - lat_A)) * sin(radians(lon_B - lon_A))

# 计算AB两点间在经度和纬度方向上的距离
delta_lat = radians(lat_B - lat_A)
delta_lon = radians(lon_B - lon_A)
a = sin(delta_lat / 2) ** 2 + cos(radians(lat_A)) * cos(radians(lat_B)) * sin(delta_lon / 2) ** 2
c = 2 * atan2(sqrt(a), sqrt(1 - a))
distance_lat = R * c

a = sin(delta_lon / 2) ** 2 + cos(radians(lat_A)) * cos(radians(lat_B)) * sin(delta_lat / 2) ** 2
c = 2 * atan2(sqrt(a), sqrt(1 - a))
distance_lon = R * c

# 将距离转换为以米为单位的距离
x_b = distance_lon
y_b = distance_lat

print("A点的XY坐标：", x_a, y_a)
print("B点的XY坐标：", x_b, y_b)
```

## 函数

传入左上角经纬度坐标（作为原点），再传入一个右下方的经纬度坐标，转换成以 米 为单位的坐标

```python
from math import sin, cos, sqrt, atan2, radians


def coordinate_transform(origin_lat,origin_lon,other_lat,other_lon):
    '''
    origin:左上角经纬度坐标 原点
    other：右下方经纬度坐标
    '''
    # 地球半径，单位米
    R = 6371000

    # A点经纬度坐标
    origin_lat = 55.75261
    origin_lon = 12.61431


    # 计算A点的经纬度对应的XY坐标
    x_a = 0
    y_a = 0

    # 计算B点的经纬度对应的XY坐标
    x_b = R * cos(radians(other_lat - origin_lat)) * cos(radians(other_lon - origin_lat))
    y_b = R * cos(radians(other_lat - origin_lon)) * sin(radians(other_lon - origin_lon))

    # 计算AB两点间在经度和纬度方向上的距离
    delta_lat = radians(other_lat - origin_lat)
    delta_lon = radians(other_lon - origin_lon)
    a = sin(delta_lat / 2) ** 2 + cos(radians(origin_lat)) * cos(radians(other_lat)) * sin(delta_lon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance_lat = R * c

    a = sin(delta_lon / 2) ** 2 + cos(radians(origin_lat)) * cos(radians(other_lat)) * sin(delta_lat / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance_lon = R * c

    # 将距离转换为以米为单位的距离
    x_b = distance_lon
    y_b = distance_lat
    
    return x_b,y_b
    
    
coordinate_transform(55.75261,12.61431,55.75262,12.61430)
```
