## train

```python
import pandas as pd

# 读取csv文件
df = pd.read_csv('05_aisdk_2023_04_13_train.csv')

# 读取的数据根据TIME和MMSI列进行从小到大排序
df.sort_values(by=['Timestamp','MMSI'], inplace=True)

# 排序后的数据保存在一个新的csv文件
df.to_csv('07_aisdk_2023_04_13_train.csv', index=False)

dict(df['Timestamp'].value_counts(sort=True))
```
