```python
import pandas as pd
import folium
import random

# 读取CSV文件
df = pd.read_csv('05_aisdk_2023_02_08_141_01.csv', usecols=['Timestamp', 'MMSI', 'Latitude', 'Longitude'])

# 根据MMSI进行分组
grouped = df.groupby('MMSI')

# 创建地图
m = folium.Map(location=[df['Latitude'].mean(), df['Longitude'].mean()], zoom_start=10)

# 为每个组添加标记
for name, group in grouped:
    color = '#' + ''.join([random.choice('0123456789ABCDEF') for j in range(6)])
    for lat, lon, in zip(group['Latitude'], group['Longitude']):
        folium.Marker(location=[lat, lon], icon=folium.Icon(color=color)).add_to(m)

# 保存地图
m.save('06_aisdk_2023_02_08_141_01.html')
```
